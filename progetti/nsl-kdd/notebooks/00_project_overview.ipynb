# NSL-KDD Intrusion Detection Benchmark

Benchmark di tesi concluso. Sei modelli a cinque famiglie e un esperimento binario finale con DLHA.

## Valutazione

Dopo la validation 80/20 interna, i sei modelli sono stati addestrati sul 100% di KDDTrain+ e valutati su KDDTest+. Il MLP ha ottenuto il macro F1 più alto (0,5469). Il suo recall su R2L è 10,71% e su U2R è 8,50%.

## Attacchi assenti dal training

Sui 3.750 casi di 17 tipi assenti, il MLP ha rilevato il 48,69% degli attacchi con il 7,32% di falsi allarmi sui normali. La reimplementazione DLHA di Wisanwanichthan e Thammawichai (2021) ha rilevato in media il 90,68% con il 16,33% di falsi allarmi. Queste misure indicano la presenza di un allarme, senza richiedere che la famiglia sia corretta.
